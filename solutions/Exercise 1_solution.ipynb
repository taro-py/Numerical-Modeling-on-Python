```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
#this function computes the orbital period in years
#the radius r should be given in AU
#the mass of the star M is given in in solar masses
def kepler(r , M):
    return (r**3/(M))**(1/2)
```

```python
r = np.linspace(0, 2.5, 200) #a mash for the radius

ySolar = kepler(r, 1.0)
yVega  = kepler(r, 2.0)
```

```python
plt.plot(r, ySolar, label = 'solar system')
plt.plot(r, yVega, label = 'Vega')
plt.xlabel("Distance in AU")
plt.ylabel("Period in years")


plt.axvline(1.0, 0, 10, label = "earth" , linestyle = 'dotted')
plt.grid(True)
plt.legend()
plt.savefig('week1_lab-kepler.pdf')
```
