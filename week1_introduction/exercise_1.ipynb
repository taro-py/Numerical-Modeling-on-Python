```python
import numpy as np

import matplotlib.pyplot as plt

##Part 1: Creating kepler(r,M) function to return the orbital perido T
def kepler(r, M):

    T_squared = ((r**3) / M)

    T = np.sqrt(T_squared)

    return T

#Part2: We generate random values for r on a vector called r_values to make the Plot
r_values = np.linspace(0.1, 10, 100)

#kepler function applied on our Solar System
T_solar = kepler(r_values, 1.0)

#kepler function applied on Vega Star
T_vega = kepler(r_values, 2.0)

#Making the code for the plot, we elaborate T as a function or r_values
plt.plot(r_values, T_solar, color="blue", label="Solar System")
plt.plot(r_values, T_vega, color="red", label="Vega Star")
plt.xlabel("r")
plt.ylabel("T")
plt.legend()

```
