```python
import numpy as np
import time 


#Creating a random 10x10 Matrix
A = np.random.rand(10, 10)

#Creating 10 element random vector
b = np.random.rand(10)

#Creating the timer to measure the time taken to solve the linear equation
start = time.time()

#Solving the linear equation Ax = b for x
x = np.linalg.solve(A, b)

#ending the timer
end = time.time()

print("Solution: \n", x)
print("Elapsed time:", end - start, "seconds")
```
