```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Part 1: Creating the Array's
t = np.array([0.3, 0.36, 0.94, 0.96, 0.98, 1.8, 1.86, 1.98])
x = np.array([0.81, -1.02, -0.05, -0.75, -1.09, -1.02, 0.08, 1.04])
```

```python
#Part 2: We plot both arrays
plt.scatter(t, x, color="black", label="data points")
plt.xlabel("Time t (s)")
plt.ylabel("Position x (cm)")
plt.legend()
plt.show()
```

```python
#Part 3: We'll make the Vandermonde Matrix (A) and the Vector (Y)

n = len(t)

A = np.zeros((n,n))

for i in range (n):
    for j in range (n):
            A[i,j] = t[i] ** j

y = np.array(x)
```

```python
#Part 4: We are going to solve the Matrix equiation using it as a linear equation

c = np.linalg.solve(A, y)

print("Coeficients of c:\t ", c)
```

```python
#Part 5: Function Polyn() made to evaluate the Polynomial given the coeficients

def Polyn(c, t):
    p = 0
    for i, cf in enumerate(c):
        p += cf * (t ** i)
    
    return p
```

```python
#Part 6: We declare an array with 500 entries for t_values and a second array with the values of the evalutaed polynimal

#we create the array 
t_plot = np.linspace(t.min(), t.max(), 500)

#we call the function to evaluate the polynomial
x_plot = Polyn(c, t_plot)
```
